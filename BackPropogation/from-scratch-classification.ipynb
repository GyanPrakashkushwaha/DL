```python
df = pd.DataFrame([[8,8,4],[7,9,5],[6,10,6],[5,12,7]], columns=['cgpa', 'profile_score', 'lpa'])
```

```python
def initialize_params(layer_dims):
    np.random.seed(3)
    params = {}
    L = len(layer_dims)
    for l in range(1, L):
        params["W" + str(l)] = np.ones((layer_dims[l-1], layer_dims[l]))*0.1
        params["b" + str(l)] = np.zeros((layer_dims[l], 1))
        
    return params
```

```python
initialize_params([2, 2, 1])
```

```python
# calculate output of a given neuron
def linear_forward(A_prev, W, b):
    Z = np.dot(W.T, A_prev) + b
    return Z
```

```python
# Forward Propagation algo
def L_layer_forward(X, parameters):
    A = X
    L = len(parameters) //2
    
    for l in range(1, L+1):
        A_prev = A 
        Wl = parameters['W' + str(l)]
        bl = parameters['b' + str(l)]
        print("A"+str(l-1)+": ", A_prev)
        print("W"+str(l)+": ", Wl)
        print("b"+str(l)+": ", bl)
        print("--"*20)

        A = linear_forward(A_prev, Wl, bl)
        print("A"+str(l)+": ", A)
        print("**"*20)
          
    return A,A_prev
```

```python
X = df[["cgpa", "profile_score"]].values[0].reshape(2,1)
y = df[["lpa"]].values[0]
# X, y

params = initialize_params([2, 2, 1])
y_hat, A_prev = L_layer_forward(X, params)
```

```python
# Update the parameters using the update rule.
def update_parameters(parameters, y, y_hat, A1, X, lr):
    loss = 2*(y - y_hat)
    lr_update = lr*loss
    
    parameters["W2"][0][0] = parameters["W2"][0][0] + lr_update*A1[0][0]
    parameters["W2"][1][0] = parameters["W2"][1][0] + lr_update*A1[1][0]
    parameters["b2"][0][0] = parameters["b2"][0][0] + lr_update
    
    parameters['W1'][0][0] = parameters['W1'][0][0] + (lr_update*parameters['W2'][0][0]*X[0][0])
    parameters['W1'][0][1] = parameters['W1'][0][1] + (lr_update*parameters['W2'][0][0]*X[1][0])
    parameters['b1'][0][0] = parameters['b1'][0][0] + (lr_update*parameters['W2'][0][0])

    parameters['W1'][1][0] = parameters['W1'][1][0] + (lr_update*parameters['W2'][1][0]*X[0][0])
    parameters['W1'][1][1] = parameters['W1'][1][1] + (lr_update*parameters['W2'][1][0]*X[1][0])
    parameters['b1'][1][0] = parameters['b1'][1][0] + (lr_update*parameters['W2'][1][0])
    
```

```python
update_parameters(params, y, y_hat[0][0], A_prev, X, 0.001)
```

```python
params = initialize_params([2, 2, 2])
epochs = 5
for i in range(epochs):
    loss = []
    for j in range(df.shape[0]):
        X = df[["cgpa", "profile_score"]].values[1].reshape(2,1)
        y = df[["lpa"]].values[1]
        y_hat, A_prev = L_layer_forward(X, params)
        update_parameters(params, y, y_hat[0][0], A_prev, X, 0.001)
        loss.append((y-y_hat[0][0])**2)
    print('Epoch - ',i+1,'Loss - ',np.array(loss).mean())
    
```
